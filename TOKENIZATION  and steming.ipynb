```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
corpus = "i am nirmal. i am a student"
documents = sent_tokenize(corpus)
```

```python
print(documents)
```

```python
for sent in documents:
    print(sent)
```

```python
from nltk.tokenize import word_tokenize
words  = word_tokenize(corpus)
print(words)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
token = TreebankWordTokenizer()
token.tokenize(corpus)
```

```python
#classification problem 
#comments on the product is positive or negitive
#reviews eat,eating,eaten,going,gone,goes,go

words = ["writing","writes","eaten","eat","eats","eating","programming","programs"]
```

```python
from nltk.stem import PorterStemmer
stems = PorterStemmer()
```

for word in words:
    print(word,stems.stem(word))

```python
stems.stem("sit")
```

```python
#regexp stemmer
from nltk.stem import RegexpStemmer
reg = RegexpStemmer('ing|s$|e$|able$',min=4)
reg.stem("eating")
```

```python
reg.stem("ingeating")
```

```python
#snow ball stemmer

from nltk.stem import SnowballStemmer
snow = SnowballStemmer('english')
```

```python
for word in words:
   print( snow.stem(word))
```
